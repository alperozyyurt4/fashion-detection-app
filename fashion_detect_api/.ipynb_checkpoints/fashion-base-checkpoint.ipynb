```python
import os

# Resim dosyalarının bulunduğu klasör
image_folder_path = '/Users/alperozyurt/Desktop/algoritma-proje/fashion-dataset/images'

# Yalnızca .jpg uzantılı dosyaları listele
image_files = [file for file in os.listdir(image_folder_path) if file.endswith('.jpg')]

# Geçerli resim dosyalarının sayısını kontrol et
print(f"Geçerli .jpg dosyası sayısı: {len(image_files)}")
```

```python
import pandas as pd

# CSV dosyasını oku
csv_path = '/Users/alperozyurt/Desktop/algoritma-proje/fashion-dataset/styles.csv'
styles_df = pd.read_csv(csv_path, on_bad_lines='skip')

# Resim dosyalarını sayfada gösterelim
print(f"styles.csv dosyasındaki satır sayısı: {len(styles_df)}")

# Resim dosyalarını ve CSV satırlarını eşitlemek için, fazla dosyaları atacağız
image_files = image_files[:len(styles_df)]

# Şimdi resim dosyalarını dataframe'e ekleyelim
styles_df['image'] = image_files

# Güncellenmiş dataframe'i kontrol edelim
print(styles_df.head())
```

```python
# Eğer image_files daha kısa ise, eksik olanları NaN ile dolduralım
styles_df['image'] = pd.Series(image_files)

# Eksik değerleri NaN ile dolduralım (resim dosyası olmayanlara)
styles_df['image'] = styles_df['image'].fillna(value='NaN')

# Güncellenmiş dataframe'i kontrol edelim
print(styles_df.head())
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# İlk 10 eşleşen resmi görüntüleyelim
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

# Eşleşen resimleri döngü ile gösterelim
for idx, ax in enumerate(axes.flat):
    # Resmin dosya adı
    image_file = styles_df['image'][idx]
    
    if image_file != 'NaN':  # NaN olmayan resimleri göstereceğiz
        img_path = os.path.join(image_folder_path, image_file)
        img = mpimg.imread(img_path)
        ax.imshow(img)
        ax.set_title(f"Image {image_file}")
        ax.axis('off')
    else:
        ax.set_title("No Image")
        ax.axis('off')

plt.tight_layout()
plt.show()

# NaN olan resimleri ve kaç tane olduğunu gösterelim
nan_count = styles_df['image'].isna().sum()
print(f"\nEksik olan resimlerin sayısı: {nan_count}")

# Eksik resimlerin dosya isimlerini ve ilgili satırları gösterelim
nan_images = styles_df[styles_df['image'] == 'NaN']
print("\nEksik resimler (NaN olanlar):")
print(nan_images[['id', 'image']])

# Kullanılmayan resimleri listeleyelim (resim dosyaları mevcut fakat CSV'de yer almayanlar)
unused_images = set(image_files) - set(styles_df['image'].dropna())
print("\nKullanılmayan resimler:")
print(unused_images)
```

```python
# Resim dosyalarının var olduğu satırlara karşılık gelen 'id' satırlarını alalım
missing_ids = set(styles_df[~styles_df['image'].isin(image_files)]['id'])

# Fazla olan satırları yazdıralım
print(f"Fazla olan satırların id'leri: {missing_ids}")
```
