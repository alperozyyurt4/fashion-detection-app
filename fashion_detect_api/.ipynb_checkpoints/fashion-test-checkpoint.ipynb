```python
from tensorflow.keras.models import load_model

# Modeli yükleme
model = load_model('model.h5')  # Kendi model dosyanızın yolu
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np

# Test için bir görüntü yolu
image_path = "/Users/alperozyurt/Desktop/algoritma-proje/fashion-dataset/test1.jpg"  # Test etmek istediğiniz görüntü

# Görüntüyü yükleme ve ön işleme
IMG_SIZE = (128, 128)  # Model eğitiminde kullandığınız boyutlar
image = load_img(image_path, target_size=IMG_SIZE)  # Görüntüyü yükle ve yeniden boyutlandır
image_array = img_to_array(image)  # Görüntüyü numpy array'e çevir
image_array = np.expand_dims(image_array, axis=0)  # Batch boyutunu ekle
image_array = image_array / 255.0  # Normalizasyon (Modeli eğitimde yaptığınız gibi)
```

```python
# Tahmin yap
predictions = model.predict(image_array)

# Tahmin edilen sınıf
predicted_class = np.argmax(predictions, axis=1)[0]  # En yüksek olasılığa sahip sınıf
print(f"Tahmin Edilen Sınıf: {predicted_class}")
```
